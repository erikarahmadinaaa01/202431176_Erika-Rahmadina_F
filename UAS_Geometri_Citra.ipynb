## Import Library

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Menampilkan Citra Asli

```python
img = cv2.imread("Foto Erika.jpeg")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Citra Asli")
plt.axis("on")
plt.show()
```

## Rotated

```python
img = cv2.imread("Foto Erika.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

(h, w) = img.shape[:2]
center = (w//2, h//2)

M = cv2.getRotationMatrix2D(center, 45, 1.0)

rotated = cv2.warpAffine(img, M, (w, h))

plt.imshow(rotated)
plt.title("After Rotated")
plt.axis("on")
plt.show()
```

## Resized

```python
img = cv2.imread("Foto Erika.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

resized = cv2.resize(img, (300,400))

plt.imshow(resized)
plt.title("After Resized")
plt.axis("on")
plt.show()
```

## Cropped

```python
img = cv2.imread("Foto Erika.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

crop = img[120:420, 150:350]

plt.imshow(crop)
plt.title("After Cropped")
plt.axis("on")
plt.show()
```

## Flipped

```python
img = cv2.imread("Foto Erika.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

flip = cv2.flip(img, 1)

plt.imshow(flip)
plt.title("After Flipped")
plt.axis("on")
plt.show()
```

## Translated

```python
img = cv2.imread("Foto Erika.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

(h, w) = img.shape[:2]

M = np.float32([[1, 0, 50],
                [0, 1, 30]])

translated = cv2.warpAffine(img, M, (w, h))

plt.imshow(translated)
plt.title("After Translated")
plt.axis("on")
plt.show()
```

```python
# Membaca gambar
img = cv2.imread("Foto Erika.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Rotasi
(h, w) = img.shape[:2]
center = (w//2, h//2)
M_rot = cv2.getRotationMatrix2D(center, 45, 1)
rotated = cv2.warpAffine(img, M_rot, (w, h))

# Resize
resized = cv2.resize(img, (0,0), fx=0.5, fy=0.5)

# Crop
cropped = img[120:420, 150:350]

# Flip
flipped = cv2.flip(img, 1)

# Translate
M_trans = np.float32([[1,0,50],[0,1,30]])
translated = cv2.warpAffine(img, M_trans, (w,h))

# Menampilkan semua hasil
plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.imshow(img)
plt.title("Citra Asli")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(rotated)
plt.title("After Rotated")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(resized)
plt.title("After Resized")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(cropped)
plt.title("After Cropped")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(flipped)
plt.title("After Flipped")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(translated)
plt.title("After Translated")
plt.axis("off")

plt.tight_layout()
plt.show()
```
